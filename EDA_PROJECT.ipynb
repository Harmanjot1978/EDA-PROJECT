```python
# outliers
plt.figure(figsize = (10,6))
sns.boxplot(x= df['species'])
plt.show()
```

```python
# Univariate Analysis
sns.histplot(df['species'],kde = True)
plt.title("species")
plt.show()
```

```python
# Bivariate Analysis
sns.boxplot(x = 'species',y='island', data = df)
plt.title("ISLAND VS SPECIES")
plt.show()
```

```python
sns.countplot(x = 'species',hue = 'island', data = df)
plt.title("ISLAND VS SPECIES")
plt.show()
```

```python
plt.subplot(1,2,2)
sns.boxplot(x = df['island'])
plt.title("island")
plt.show()
```

```python
# Multivariate Analysis
plt.figure(figsize =(12,6))
sns.heatmap(df.select_dtypes(include = 'number').corr(),annot = True, cmap = 'coolwarm', fmt = '.2f')
plt.title("Heatmap")
plt.show()
```

```python
df.select_dtypes(include = 'number').plot(kind = 'box',figsize = (12,6),rot = 45)
plt.show()
```
